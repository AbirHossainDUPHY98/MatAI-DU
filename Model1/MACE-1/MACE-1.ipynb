````python
from ase.io import read, write

train_atoms = read("train.xyz", index=":")
test_atoms = read("test.xyz", index=":")

removed_count = 0
for atoms in train_atoms + test_atoms:
    if "field_scalar" in atoms.arrays:
        atoms.arrays.pop("field_scalar")
        removed_count += 1

print(f"Stripped field_scalar from {removed_count} configs")

write("train_clean.xyz", train_atoms, format="extxyz")
write("test_clean.xyz", test_atoms, format="extxyz")
````

````python
from ase.io import read, write

train_atoms = read("train.xyz", index=":")
test_atoms = read("test.xyz", index=":")

for atoms in train_atoms + test_atoms:
    if "field_scalar" in atoms.arrays:
        del atoms.arrays["field_scalar"]

    # forces live on the calculator, not atoms.arrays -- pull them out first
    forces = atoms.get_forces()
    energy = atoms.get_potential_energy()

    # detach the calculator so nothing re-derives 'forces' key on write
    atoms.calc = None

    # add force as a proper per-atom array, and stash energy back into info
    atoms.new_array("force", forces)
    atoms.info["energy"] = energy

print("Arrays now:", list(train_atoms[0].arrays.keys()))
print("Info now:", list(train_atoms[0].info.keys()))

write("train_clean.xyz", train_atoms, format="extxyz")
write("test_clean.xyz", test_atoms, format="extxyz")
````

**Top-level group name:** H64O32Pt64 — this is dpdata's naming convention: it labels each distinct chemical formula as its own group. Since all 80 frames share the same composition (64 Pt, 32 O, 64 H — matching what we identified earlier as the Pt slab + water system), everything lives under one group. If your dataset had frames with varying atom counts/compositions, you'd see multiple sibling groups here.

**Inside** set.000/ (DeepMD's convention for a "set" — a batch/shard of frames with identical atom count and ordering):

- box.npy: shape (80, 9) — the 3×3 lattice/cell matrix flattened to 9 values, one row per frame. Matches your 80 training configs.
- coord.npy: shape (80, 480) — atomic positions, flattened. 480 = 160 atoms × 3 (x,y,z). One row per frame.
- energy.npy: shape (80, 1) — total energy per frame, matches the energy= values we saw in the raw xyz.
- force.npy: shape (80, 480) — forces, same flattening scheme as coords: 160 atoms × 3.

**Outside** set.000/, at the H64O32Pt64/ level:

- type.raw: shape (160,) — one integer per atom (not per frame), giving each atom's type index (0, 1, or 2, mapping into type_map.raw). This is fixed across all frames in this group since atom ordering matches.
- type_map.raw: shape (3,), dtype |S2 (byte-strings, max 2 chars) — the actual element symbols, e.g. [b'H', b'O', b'Pt'], defining what type index 0/1/2 means.

- **field_scalar** — the PE_MACE field-conditioning value, not representable in DeepMD's schema
- **stress** — DeepMD/dpdata can store stress/virial too, but we didn't request it and the header we wrote didn't include it (worth adding back if you need it — dpdata supports a virials.npy array if --stress data is present in the source)
- **dipole, magmom** — similarly dropped

````python
with h5py.File("/content/PE_MACE_train/train_0.h5", "r") as f:
    def print_tree(name, indent=0):
        item = f[name]
        if isinstance(item, h5py.Group):
            print("  " * indent + f"{name.split('/')[-1]}/")
            for k in item.keys():
                print_tree(f"{name}/{k}" if name else k, indent + 1)
        else:
            val = item[()]
            preview = val if (hasattr(val, "size") and val.size < 10) else f"shape={item.shape}"
            print("  " * indent + f"{name.split('/')[-1]}: {preview}")
    for top_key in f.keys():
        print_tree(top_key)
````

# MACE Training & Evaluation — Working Reference

Distilled from a working session on MACE 0.3.17. Covers both the direct-XYZ
path and the HDF5-preprocessing path, plus every gotcha that actually broke
something.

---

## 0. Data format requirements (apply to BOTH paths)

Your extended-XYZ file's `Properties=` header must contain **exactly**:
```
Properties=species:S:1:pos:R:3:forces:R:3
```
- Use `forces` (plural) — this is what MACE's own reader expects.
- Any extra per-atom field not in {`species`, `pos`, `forces`} (e.g. a custom
  `field_scalar` column) is fine for MACE's own tools, but will **hard-crash**
  dpdata or QUIP/GAP-style parsers. Strip such columns before using anything
  outside the MACE ecosystem.
- Global key=value line must include `energy=...`. Don't rely on `energy`
  as the key name long-term — ASE/MACE warn that `energy`/`forces` are
  "unsafe" going forward; prefer `REF_energy`/`REF_forces` and pass
  `--energy_key=REF_energy --forces_key=REF_forces` if starting fresh.
- Download raw files with `raw.githubusercontent.com`, never a GitHub
  `blob/` URL — the latter downloads an HTML page, not the data.

---

## 1. Direct XYZ training (simplest path, good for quick tests)

```bash
mace_run_train \
    --name="RUN_NAME" \
    --train_file="train.xyz" \
    --test_file="test.xyz" \
    --valid_fraction=0.1 \
    --energy_key="energy" \
    --forces_key="forces" \
    --E0s="average" \
    --model="MACE" \
    --hidden_irreps="128x0e + 128x1o" \
    --r_max=5.0 \
    --batch_size=4 \
    --max_num_epochs=500 \
    --device=cuda \
    --wandb --wandb_project="PROJECT_NAME" --wandb_name="RUN_NAME"
```

**Must-dos:**
- Always include `--test_file` if you want real held-out test metrics —
  without it, MACE only reports train/valid (validation is just a random
  10% split of train, not a true test set).
- `--max_num_epochs=5` is a smoke test, not a real run. Use hundreds+ and
  watch wandb/loss curves for a plateau before trusting results.
- No trailing spaces after line-continuation `\` — a stray space causes
  `unrecognized arguments: \`.
- In Jupyter/Colab, prefix shell commands with `!`.

**Evaluating an already-trained model on new data (no retraining):**
```bash
mace_eval_configs \
    --configs="test.xyz" \
    --model="checkpoints/RUN_NAME_run-123.model" \
    --output="test_eval.xyz" \
    --device=cuda \
    --batch_size=4
```
- No `--energy_key`/`--forces_key` here — not accepted by this tool.
- If CUDA OOM: lower `--batch_size` (try 4, 2, 1) before anything else.

---

## 2. HDF5 preprocessing + training (for larger/repeated-use datasets)

### Step 1 — Preprocess (module has no console-script entry point in 0.3.17):
```bash
python -m mace.cli.preprocess_data \
    --train_file="train.xyz" \
    --valid_fraction=0.1 \
    --test_file="test.xyz" \
    --h5_prefix="PREFIX_" \
    --r_max=5.0 \
    --energy_key="energy" \
    --forces_key="forces" \
    --atomic_numbers="[Z1, Z2, Z3]" \
    --compute_statistics \
    --E0s="average" \
    --num_process=1 \
    --seed=123
```
**Gotchas:**
- `--work_dir` is effectively ignored in this version — output lands in the
  current directory as `PREFIX_train/`, `PREFIX_val/`, `PREFIX_test/`,
  `PREFIX_statistics.json`. Locate with `find / -iname "PREFIX_*"` if unsure.
- `--atomic_numbers` must be given explicitly if E0s aren't in the file.

### Step 2 — Fix the statistics.json numpy-serialization bug:
The tool writes `np.float64(...)` wrapped inside a JSON *string*, which
`ast.literal_eval` (used by `mace_run_train`) cannot parse. Patch before use:
```python
import re
with open("PREFIX_statistics.json") as f:
    content = f.read()
content = re.sub(r"np\.float64\(([^)]+)\)", r"\1", content)
with open("PREFIX_statistics_fixed.json", "w") as f:
    f.write(content)
```

### Step 3 — Patch two known bugs in `mace/cli/run_train.py` (0.3.17 only, check if fixed in newer versions):
```python
path = "/usr/local/lib/python3.12/dist-packages/mace/cli/run_train.py"
with open(path) as f: c = f.read()
c = c.replace(
    'test_folders = glob(head_config.test_dir + "/*")',
    'test_folders = glob.glob(head_config.test_dir + "/*")')
c = c.replace(
    'for folder in test_folders:\n                    name = os.path.splitext(os.path.basename(test_file))[0]',
    'for folder in test_folders:\n                    name = os.path.splitext(os.path.basename(folder))[0]')
with open(path, "w") as f: f.write(c)
```
(Bug 1: `glob` imported as module but called as function. Bug 2: stale loop
variable `test_file` used instead of `folder`.)

### Step 4 — Train from HDF5:
```bash
mace_run_train \
    --name="RUN_NAME" \
    --train_file="/path/PREFIX_train" \
    --valid_file="/path/PREFIX_val" \
    --test_file="/path/PREFIX_test/Default__0.h5" \
    --statistics_file="/path/PREFIX_statistics_fixed.json" \
    --model="MACE" \
    --hidden_irreps="128x0e + 128x1o" \
    --r_max=5.0 \
    --batch_size=4 \
    --max_num_epochs=500 \
    --device=cuda \
    --wandb --wandb_project="PROJECT_NAME" --wandb_name="RUN_NAME"
```
**Must-dos:**
- `--train_file` / `--valid_file` point at **folders** of shards.
- `--test_file` points at the actual **single `.h5` file** (e.g.
  `Default__0.h5`) directly — NOT `--test_dir` +
  `--multi_processed_test=True`, which is only for a directory containing
  *subfolders* of shards (a different, sharded-multi-file layout you likely
  won't have unless you used `--num_process` > 1 during preprocessing to
  generate multiple shards, or an actual multi-shard dataset).
- With `--statistics_file` given, drop `--E0s`/`--energy_key`/`--forces_key`
  — they're already baked into the HDF5 labels.
- Test shard filenames are prefixed by head name + double underscore, e.g.
  `Default__0.h5` — check via `find PREFIX_test -type f` rather than
  assuming a name.

---

## 3. Do NOT use dpdata for MACE HDF5 (dead end, confirmed this session)

dpdata produces DeepMD-kit-schema HDF5 (`set.000/coord.npy` style), which
`mace_run_train` cannot read — the internal key names and layout are
fundamentally different from MACE's own `config_batch_0/config_N/...`
schema. Only use dpdata if you're actually training DeepMD-kit, not MACE.
If you need MACE HDF5, always go through `mace.cli.preprocess_data`.

---

## 4. General sanity-check habits worth keeping

- After any format conversion, spot-check one energy value against the
  original raw XYZ (`grep energy= file.xyz | head -1`) to confirm no
  silent data loss/reordering.
- After training, always check the RESULTS section actually printed an
  `Evaluating test_Default ...` line — its absence means the test set was
  silently skipped, not that it passed with zero error.
- 5 epochs / ~80 configs is a "does the pipeline run" smoke test only.
  Expect RMSE_E in the hundreds of meV/atom at that scale — not a real
  quality signal. For real DFT data, plan for hundreds–thousands of epochs
  and a dataset several times larger before trusting reported errors.
